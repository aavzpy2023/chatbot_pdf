```python
import json
import re


def process_section(section):
    lines = section.strip().split("\n")
    data = {}
    current_key = None

    for line in lines:
        line = line.strip()
        if not line:
            continue  # Ignorar líneas en blanco
        if line.startswith("Sct."):
            # Extraer la clave y el valor inicial
            key = line[4:].split(":", 1)[0].strip()
            value = line[4:].split(":", 1)[1].strip() if ":" in line[4:] else ""
            if key in [
                "Variantes de Preguntas",
                "Pasos a Seguir",
                "Resultados Esperados",
                "Escenarios Posibles y Casos Especiales",
            ]:
                current_key = key
                data[key] = [value] if value else []
            else:
                data[key] = value
        elif current_key:
            data[current_key].append(line)

    return data


def convert_text_to_json(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()

    sections = re.split(r"ID: ", content)[1:]
    result = {}

    for section in sections:
        id_match = re.match(r"(\S+)", section)
        if id_match:
            id_value = id_match.group(1)
            section_data = process_section(section)
            result[id_value] = section_data

    return result
```

```python
import requests
import numpy as np
import json

# URL de la API local
API_URL = 'http://localhost:5000/generate-embeddings/'

def get_embeddings(texts):
    headers = {
        'Content-Type': 'application/json'
    }
    data = {
        "texts": texts
    }
    response = requests.post(API_URL, headers=headers, data=json.dumps(data))

    if response.status_code == 200:
        return response.json()['embeddings']
    else:
        raise Exception(f"Error: {response.status_code}, {response.text}")
```

```python
from pymilvus import MilvusClient, DataType


client.drop_collection(collection_name="Sarasola")

# Crear el esquema
schema = MilvusClient.create_schema(
    auto_id=False,  # No usamos ID automático
    enable_dynamic_field=False  # Deshabilitar campos dinámicos
)

# Agregar campos al esquema
schema.add_field(field_name="q_id", datatype=DataType.VARCHAR, is_primary=True, max_length=64)
schema.add_field(field_name="q_vector", datatype=DataType.FLOAT_VECTOR, dim=768)  # Dimensión del vector
schema.add_field(field_name="q_question", datatype=DataType.VARCHAR, max_length=512)

# Crear la colección
client.create_collection(
    collection_name="Sarasola",
    schema=schema
)

# Crear un índice para el campo vectorial
index_params = {
    "metric_type": "L2",  # Distancia euclidiana
    "index_type": "IVF_FLAT",
    "params": {"nlist": 128}
}
print(index_params)
print(type(index_params))
```
