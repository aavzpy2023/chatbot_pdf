```python
import re

def process_file(file_path):
    # Leer el archivo
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()

    # Expresión regular para capturar IDs y su contenido
    pattern = r"(ID:\s*[A-Za-z0-9_-]+_[\w-]+_P\d+)(.*?)(?=ID:\s*[A-Za-z0-9_-]+_[\w-]+_P\d+|\Z)"

    # Encontrar todas las coincidencias
    matches = re.findall(pattern, content, re.DOTALL)

    # Crear un diccionario con los IDs como claves y el contenido como valores
    result = {}
    for match in matches:
        question_id = match[0].strip()  # ID de la pregunta
        question_content = match[1].strip()  # Contenido asociado al ID
        result[question_id] = question_content

    return result


# Ruta del archivo
file_path = "./documents/mf3.txt"

# Procesar el archivo
questions_dict = process_file(file_path)

# Mostrar el diccionario resultante
for question_id, question_content in questions_dict.items():
    print(f"ID: {question_id}")
    print(f"Contenido:\n{question_content}\n{'-' * 50}")
```

```python
import json
import re


def process_section(section):
    lines = section.strip().split("\n")
    data = {}
    current_key = None

    for line in lines:
        line = line.strip()
        if not line:
            continue  # Ignorar líneas en blanco
        if line.startswith("Sct."):
            # Extraer la clave y el valor inicial
            key = line[4:].split(":", 1)[0].strip()
            value = line[4:].split(":", 1)[1].strip() if ":" in line[4:] else ""
            if key in [
                "Variantes de Preguntas",
                "Pasos a Seguir",
                "Resultados Esperados",
                "Escenarios Posibles y Casos Especiales",
            ]:
                current_key = key
                data[key] = [value] if value else []
            else:
                data[key] = value
        elif current_key:
            data[current_key].append(line)

    return data


def convert_text_to_json(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()

    sections = re.split(r"ID: ", content)[1:]
    
    # print("SECTION:",list(sections)[-1])
    result = {}

    for section in sections:
        id_match = re.match(r"(\S+)", section)
        if id_match:
            id_value = id_match.group(1)
            section_data = process_section(section)
            result[id_value] = section_data

    return result
```

```python
import requests
import numpy as np
import json

# URL de la API local
API_URL = 'http://localhost:5000/generate-embeddings/'

def get_embeddings(texts):
    headers = {
        'Content-Type': 'application/json'
    }
    data = {
        "texts": texts
    }
    response = requests.post(API_URL, headers=headers, data=json.dumps(data))

    if response.status_code == 200:
        return response.json()['embeddings']
    else:
        raise Exception(f"Error: {response.status_code}, {response.text}")
```

```python
def get_question_contents(questions_id: list):
    file_path = "./documents/mf3.txt"  # Leer el archivo
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()

    # Expresión regular para capturar IDs y su contenido
    pattern = r"(ID:\s*[A-Za-z0-9_-]+_[\w-]+_P\d+)(.*?)(?=ID:\s*[A-Za-z0-9_-]+_[\w-]+_P\d+|\Z)"

    # Encontrar todas las coincidencias
    matches = re.findall(pattern, content, re.DOTALL)

    # Crear un diccionario con los IDs como claves y el contenido como valores
    result = {}
    for match in matches:
        question_id = match[0].strip()  # ID de la pregunta (incluye "ID: ")
        question_content = match[1].strip()  # Contenido asociado al ID
        result[question_id] = question_content

    # Ajustar los IDs en questions_id para que coincidan con las claves del diccionario
    adjusted_questions_id = [f"ID: {c_id}" for c_id in questions_id]

    # Obtener los contenidos correspondientes a los IDs ajustados
    contents = [result.get(c_id, "") for c_id in adjusted_questions_id]

    return contents
```
